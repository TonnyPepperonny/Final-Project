# Чеклист для сдачи

1. Самое главное - есть все необходимые утилитарные файлы, которые позволяют просто склонировать репозиторий, запустить 1 скрипт и получить результат (сабмит и метрику лучшей модели или таблицу результатов):

- config, меняя поля которого можно менять гиперпараметры, задачи, модели и т.д.
- main.py, запуская который воспроизводится основной пайплайн подгрузки данных, препроцессинга и обучения модели с кросс-валидацией
- requirements.txt / makefile / pyproject.toml, который запускается для установки окружения

2. Есть отдельный ipynb с EDA (в нем может быть не только EDA, можно хоть весь цикл обучения модели и эксперименты) - тут хочется видеть осмысленные аналитические выводы в виде текста и хорошие визуализации

3. Есть емко и информативно описанный readme файл, в котором основные результаты проекта и инструкция по запуску репозитория

4. Если для написания кода использовались LLM-ки, то код нужно почистить от AI-слопа и странных комментов

5. В целом по коду нужны текстовые комментарии для сложных логик и вспомогательных функций, функции и классы имеют докстринги, но комменты типа "подгружаю библиотеки" перед import'ами, пожалуй, излишни

6. В идеале повторно используемый код вынесен в отдельные функции и модули для лучшей читаемости и выполнения принципа DRY (dont repeat yourself)

7. Построена хотя бы 1 DL-модель для каждой соревы

8. Построено как минимум по 2 разных бустинга для задачи

9. Построен бейзлайн (линрег/knn/логрег или просто правило), с которым сравниваются остальные архитектуры

10. Код в целом чистый, нет больших закомменченных блоков, лишних переменных, имена переменных информативные
